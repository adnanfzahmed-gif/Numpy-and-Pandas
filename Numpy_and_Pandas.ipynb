```python
#Numpy Array Operations:

#1. Create a 1D NumPy Array:

import numpy as np

temperatures_w1 = np.array([22.5, 25.3, 20.8, 23.4, 26.1, 24.8, 21.9])
temperatures_w1
```

```python
#2. Inspection and Properties:

print(f"The shape of the array is : {temperatures_w1.shape}")
print(f"The size of the array is : {temperatures_w1.size}")
print(f"The data type of the array is : {temperatures_w1.dtype}")
```

```python
#3. Array Operations:

#Convert the temperatures_w1 from Celsius to Fahrenheit

temperatures_w1_fahrenheit = (temperatures_w1 * 9/5) + 32
temperatures_w1_fahrenheit

#Find the maximum, minimum, and mean temperatures for the week.

print(f"The maximum temperature for the week is : {np.max(temperatures_w1)}°C")
print(f"The minimum temperature for the week is : {np.min(temperatures_w1)}°C")
print(f"The mean temperature for the week is : {np.mean(temperatures_w1)}°C")

print(f"The maximum temperature for the week in Fahrenheit is : {np.max(temperatures_w1_fahrenheit)}°F")
print(f"The minimum temperature for the week in Fahrenheit is : {np.min(temperatures_w1_fahrenheit)}°F")
print(f"The mean temperature for the week in Fahrenheit is : {np.mean(temperatures_w1_fahrenheit)}°F")
```

```python
#4. Array Slicing and Indexing:

#Extract the temperatures for the first three days of the week.

temperatures_first_three_days_fahrenheit = temperatures_w1_fahrenheit[:3]
print(f"Temperatures for the first three days: {temperatures_first_three_days_fahrenheit}°F")

#Extract the temperatures for the weekend (last two days).

temperatures_weekend_fahrenheit = temperatures_w1_fahrenheit[-2:]
print(f"Temperatures for the weekend: {temperatures_weekend_fahrenheit}°F")

#Extract the temperatures for the middle three days of the week.

temperatures_middle_three_days_fahrenheit = temperatures_w1_fahrenheit[2:5]
print(f"Temperatures for the middle three days: {temperatures_middle_three_days_fahrenheit}°F")
```

```python
#5. Create a 2D Array:

#Create a 2D NumPy array named temperatures where each row represents
#temperatures for a different week:
#★ Row 1: [22.5, 25.3, 20.8, 23.4, 26.1, 24.8, 21.9] (Week 1)
#★ Row 2: [19.2, 22.5, 21.3, 24.0, 23.5, 22.8, 20.1] (Week 2)

temperatures = np.array([[22.5, 25.3, 20.8, 23.4, 26.1, 24.8, 21.9],
                         [19.2, 22.5, 21.3, 24.0, 23.5, 22.8, 20.1]])
temperatures
```

```python
#6. Inspect and Slice the 2D Array:

# Inspect and print the shape, data type, and total number of elements in the temperatures array.

print(f"Shape: {temperatures.shape}")
print(f"Data Type: {temperatures.dtype}")
print(f"Total Elements: {temperatures.size}")

# Extract the temperatures for each week and the weekends (last two days) for both weeks.

week1_temperatures = temperatures[0]
week2_temperatures = temperatures[1]

weekend_temperatures = temperatures[:, -2:]

# Last two days for both weeks

print(f"Week 1 Temperatures: {week1_temperatures}")
print(f"Week 2 Temperatures: {week2_temperatures}")
print(f"Weekend Temperatures: {weekend_temperatures}")
```

Pandas Series:

```python
#1. Creating Pandas Series:

#Create a Pandas Series named marks with the following data:
# Marks: 95, 92, 89, 85, 80
# Custom Indices: 'Rank1', 'Rank2', 'Rank3', 'Rank4', 'Rank5'

import pandas as pd

marks = pd.Series(data=[95, 92, 89, 85, 80], index=['Rank1', 'Rank2', 'Rank3', 'Rank4', 'Rank5'])
marks
```

```python
#2. Indexing and Slicing:

#Use the integer index position to access the mark of the 1st rank student.

print(f"The mark of the 1st rank student is: {marks.iloc[0]}")
```

```python
#Use the loc accessor to retrieve the marks of the top 3 ranks by specifying their index labels.
top_3_marks = marks.loc[['Rank1', 'Rank2', 'Rank3']]
top_3_marks
```

```python
#Use the iloc accessor to retrieve the mark of the 3rd-ranked student.
third_rank_mark = marks.iloc[2]
print(f"The mark of the 3rd rank student is: {third_rank_mark}")
```

```python
#Apply a boolean mask to filter the Series and retrieve the ranks where the marks are greater than 90.
high_marks = marks[marks > 90]
high_marks
```

```python
#3. Manipulating Series:

#Modify the mark of the student with the 1st rank to 100.
marks.loc['Rank1'] = 100
marks
```

```python
#Remove the entry corresponding to the last rank from the Series.
marks = marks.drop(marks.index[-1])
marks
```

```python
#Compute the CGPA by dividing each mark in the Series by 10.
cgpa = marks / 10
cgpa
```

```python
#Pandas DataFrame:

#1. Creating Pandas DataFrame:


transactions = pd.DataFrame({
    'TransactionID': [101, 102, 103, 104, 105, 106, 107, 108, 109, 110],
    'ProductCategory': ['Electronics', 'Clothing', 'Electronics', 'Furniture', 'Clothing',
'Electronics', 'Furniture', 'Clothing', 'Furniture', 'Electronics'],
    'Region': ['North', 'South', 'North', 'East', 'West', 'North', 'East', 'West', 'South', 'North'],
    'Amount': [200, 150, 300, 450, 200, 250, 300, 180, 350, 400]
})
transactions
```

```python
from google.colab import sheets
sheet = sheets.InteractiveSheet(df=transactions)
```

```python
#2. Data Exploration:

#Display the transactions DataFrame and its basic information, including head, tail, shape, column names, and data types.

dataframe=transactions
print("\nHead of the DataFrame:\n:",dataframe.head())
print("\nTail of the DataFrame:\n",dataframe.tail())
print("\nShape of the DataFrame:\n",dataframe.shape)
print("\nColumn Names of the DataFrame:\n",dataframe.columns)
print("\nData Types of the DataFrame:\n",dataframe.dtypes)
```

```python
#Display only the 'ProductCategory' and 'Amount' columns.

print("\nProduct Category and Amount columns:\n", dataframe[['ProductCategory', 'Amount']])
```

```python
print("The product category")
print(dataframe['ProductCategory'])

print("The amount")
print(dataframe['Amount'])
```

```python
#Retrieve the last 3 columns of the DataFrame using iloc or loc.

print(f"last 3 columns:\n{dataframe.iloc[:, -3:]}")
```

```python
#Filter rows where the 'Region' is 'North' and 'Amount' is greater than 200.

filtered_data = dataframe[(dataframe['Region'] == 'North') & (dataframe['Amount'] > 200)]
print("\nFiltered Data:\n", filtered_data)
```

```python
#Find the value counts for the 'ProductCategory' column.

value_counts = dataframe['ProductCategory'].value_counts()
print("\nValue Counts for ProductCategory:\n", value_counts)
```

```python
#Find the unique values in the 'Region' column.

unique_regions = dataframe['Region'].unique()
print("\nUnique Regions:\n", unique_regions)
```

```python
#Group by 'Region' and find the mean amount for each region.

mean_amount_by_region = dataframe.groupby('Region')['Amount'].mean()
print("\nMean Amount by Region:\n", mean_amount_by_region)
```

```python
#3. Manipulating the DataFrame:

#Modify the 'Amount' for the transaction with TransactionID 102 to 165.

dataframe.loc[dataframe['TransactionID'] == 102, 'Amount'] = 165
print("\nModified Data:\n", dataframe[dataframe['TransactionID'] == 102])
```

```python
#Add a new column 'Discount' by calculating 10% of the 'Amount'.

dataframe['Discount'] = dataframe['Amount'] * 0.10
print("\nDataFrame with Discount Column:\n", dataframe)
```

```python
#Remove the row with TransactionID 109.

dataframe = dataframe[dataframe['TransactionID'] != 109]
print("\nDataFrame after removing TransactionID 109:\n", dataframe)
```

```python
#Delete the 'Discount' column from the DataFrame.

dataframe.drop(columns=['Discount'], inplace=True)

dataframe
```
